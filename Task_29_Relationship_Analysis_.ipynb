```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df.head()
```

```python
print("First 5 Rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
x = df['sepal length (cm)']
y = df['petal length (cm)']
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(x, y, color='blue', marker='o', alpha=0.7)

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')

plt.title('Relationship Between Sepal Length and Petal Length')

plt.grid(True)

plt.show()
```

```python
correlation = df['sepal length (cm)'].corr(
    df['petal length (cm)']
)

print("Correlation Coefficient:", correlation)
```

```python
if correlation > 0:
    print("There is a Positive Relationship.")
elif correlation < 0:
    print("There is a Negative Relationship.")
else:
    print("There is No Linear Relationship.")
```

```python
print("OBSERVATIONS:")

print("1. There is a positive relationship between sepal length and petal length.")

print("2. As sepal length increases, petal length generally increases.")

print("3. The scatter plot shows an upward trend, indicating a positive correlation.")
```
